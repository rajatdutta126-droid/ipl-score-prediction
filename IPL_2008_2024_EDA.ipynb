```python
# ============================================================
# IPL 2008–2024 — Exploratory Data Analysis
# Match-level dataset
# ============================================================

import warnings
warnings.filterwarnings("ignore")

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# ------------------------------------------------------------
# 0. Configuration
# ------------------------------------------------------------
DATA_PATH = "matches.csv"   # Change this if your CSV is elsewhere
OUTPUT_DIR = Path("ipl_eda_outputs")
OUTPUT_DIR.mkdir(exist_ok=True)

# ------------------------------------------------------------
# 1. Load data
# ------------------------------------------------------------
df = pd.read_csv(DATA_PATH)

print("Shape:", df.shape)
display(df.head())

# Convert date
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Basic inspection
display(df.tail())
display(df.sample(min(5, len(df)), random_state=42))
display(df.dtypes.to_frame("dtype"))
display(df.describe(include="all").T)

# ------------------------------------------------------------
# 2. Dataset profile
# ------------------------------------------------------------
profile = pd.DataFrame({
    "column": df.columns,
    "dtype": df.dtypes.astype(str).values,
    "missing_count": df.isna().sum().values,
    "missing_pct": (df.isna().mean() * 100).round(2).values,
    "unique_values": df.nunique(dropna=True).values
})
display(profile)

print("Duplicate rows:", df.duplicated().sum())
print("Seasons:", df["season"].nunique())
print("Teams:", len(set(df["team1"]) | set(df["team2"])))
print("Venues:", df["venue"].nunique())
print("Cities:", df["city"].nunique())

# ------------------------------------------------------------
# 3. Data quality assessment
# ------------------------------------------------------------
missing = df.isna().sum().sort_values(ascending=False)
missing_pct = (df.isna().mean() * 100).sort_values(ascending=False)
missing_table = pd.DataFrame({
    "missing_count": missing,
    "missing_percentage": missing_pct.round(2)
})
display(missing_table)

plt.figure(figsize=(10, 5))
missing_nonzero = missing[missing > 0]
if len(missing_nonzero):
    sns.barplot(x=missing_nonzero.values, y=missing_nonzero.index)
    plt.title("Missing Values by Column")
    plt.xlabel("Number of Missing Values")
    plt.ylabel("Column")
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "01_missing_values.png", dpi=200, bbox_inches="tight")
    plt.show()
else:
    print("No missing values found.")

# Duplicate analysis
duplicate_count = df.duplicated().sum()
duplicate_pct = duplicate_count / len(df) * 100
print(f"Duplicate rows: {duplicate_count} ({duplicate_pct:.2f}%)")

# Missing winners: inspect before any decision
display(df[df["winner"].isna()][
    ["id", "season", "team1", "team2", "result", "result_margin"]
])

# Missing city inspection
display(df[df["city"].isna()][["id", "season", "venue"]].head(20))

# ------------------------------------------------------------
# 4. Cleaning decisions
# ------------------------------------------------------------
# Keep original df unchanged; create a cleaned copy.
clean = df.copy()

# Standardize team names only where historical franchise renaming is clear.
team_map = {
    "Delhi Daredevils": "Delhi Capitals",
    "Kings XI Punjab": "Punjab Kings",
    "Royal Challengers Bangalore": "Royal Challengers Bengaluru",
    "Rising Pune Supergiants": "Rising Pune Supergiant",
}

for col in ["team1", "team2", "toss_winner", "winner"]:
    clean[col] = clean[col].replace(team_map)

# Fill city from venue only for the known overseas venues where city is absent.
# This is a documented, deterministic cleaning rule based on the venue itself.
city_from_venue = {
    "Dubai International Cricket Stadium": "Dubai",
    "Sharjah Cricket Stadium": "Sharjah",
}
for venue, city in city_from_venue.items():
    clean.loc[(clean["city"].isna()) & (clean["venue"] == venue), "city"] = city

# Do NOT fill winner for no-result matches.
# Do NOT fill method; NaN means D/L was not recorded for most matches.
# Keep player_of_match missing where the match had no player award.
# Keep result_margin missing for ties/no-results.

print("Shape before cleaning:", df.shape)
print("Shape after cleaning:", clean.shape)

cleaning_summary = pd.DataFrame({
    "operation": [
        "Converted date to datetime",
        "Standardized historical franchise names",
        "Filled missing city for known Dubai/Sharjah venues",
        "Preserved no-result winners as missing",
        "Preserved conditional method field",
        "Preserved genuine missing result margins"
    ],
    "reason": [
        "Required for date-based analysis",
        "Avoids treating renamed franchises as unrelated teams",
        "City can be determined directly from the venue",
        "No winner exists for a no-result match",
        "D/L is only populated for affected matches",
        "Ties/no-results do not have a normal winning margin"
    ]
})
display(cleaning_summary)

# ------------------------------------------------------------
# 5. Univariate analysis
# ------------------------------------------------------------
numerical_cols = [
    c for c in ["result_margin", "target_runs", "target_overs"]
    if c in clean.columns
]

display(clean[numerical_cols].describe().T)

for col in numerical_cols:
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    sns.histplot(clean[col].dropna(), kde=True, ax=axes[0])
    axes[0].set_title(f"Distribution of {col}")
    axes[0].set_xlabel(col)
    sns.boxplot(x=clean[col], ax=axes[1])
    axes[1].set_title(f"Box Plot of {col}")
    axes[1].set_xlabel(col)
    plt.tight_layout()
    plt.show()

categorical_cols = [
    "season", "match_type", "toss_decision", "result", "super_over", "method"
]
for col in categorical_cols:
    if col in clean.columns:
        counts = clean[col].fillna("Missing").value_counts()
        plt.figure(figsize=(10, 5))
        sns.barplot(x=counts.values, y=counts.index)
        plt.title(f"{col.replace('_', ' ').title()} Distribution")
        plt.xlabel("Match Count")
        plt.ylabel(col.replace("_", " ").title())
        plt.tight_layout()
        plt.show()

# ------------------------------------------------------------
# 6. Season-wise analysis
# ------------------------------------------------------------
season_order = list(clean["season"].drop_duplicates())

season_counts = clean.groupby("season").size().reindex(season_order)
display(season_counts.to_frame("matches"))

plt.figure(figsize=(12, 5))
sns.lineplot(x=season_counts.index, y=season_counts.values, marker="o")
plt.title("IPL Matches by Season")
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "02_matches_by_season.png", dpi=200, bbox_inches="tight")
plt.show()

season_toss = pd.crosstab(clean["season"], clean["toss_decision"]).reindex(season_order)
display(season_toss)

season_avg_margin = clean.groupby("season")["result_margin"].mean().reindex(season_order)
plt.figure(figsize=(12, 5))
sns.lineplot(x=season_avg_margin.index, y=season_avg_margin.values, marker="o")
plt.title("Average Winning Margin by Season")
plt.xlabel("Season")
plt.ylabel("Average Margin")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "03_average_margin_by_season.png", dpi=200, bbox_inches="tight")
plt.show()

# ------------------------------------------------------------
# 7. Team performance
# ------------------------------------------------------------
teams = sorted(set(clean["team1"]) | set(clean["team2"]))

matches_played = pd.concat([
    clean["team1"].value_counts(),
    clean["team2"].value_counts()
], axis=1).fillna(0).sum(axis=1).sort_values(ascending=False)

wins = clean["winner"].value_counts().reindex(teams).fillna(0)
team_summary = pd.DataFrame({
    "matches_played": matches_played.reindex(teams).fillna(0).astype(int),
    "wins": wins.astype(int)
})
team_summary["win_percentage"] = (
    team_summary["wins"] / team_summary["matches_played"] * 100
).round(2)

display(team_summary.sort_values("wins", ascending=False))

# Matches played
plt.figure(figsize=(11, 7))
x = matches_played.sort_values()
sns.barplot(x=x.values, y=x.index)
plt.title("Matches Played by Team")
plt.xlabel("Matches Played")
plt.ylabel("Team")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "04_matches_played_by_team.png", dpi=200, bbox_inches="tight")
plt.show()

# Wins
win_sorted = wins.sort_values()
plt.figure(figsize=(11, 7))
sns.barplot(x=win_sorted.values, y=win_sorted.index)
plt.title("Match Wins by Team")
plt.xlabel("Wins")
plt.ylabel("Team")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "05_wins_by_team.png", dpi=200, bbox_inches="tight")
plt.show()

# Win percentage
wp = team_summary["win_percentage"].sort_values()
plt.figure(figsize=(11, 7))
sns.barplot(x=wp.values, y=wp.index)
plt.title("Win Percentage by Team")
plt.xlabel("Win Percentage")
plt.ylabel("Team")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "06_win_percentage_by_team.png", dpi=200, bbox_inches="tight")
plt.show()

# Season x team wins heatmap
team_season_wins = pd.crosstab(clean["winner"], clean["season"]).reindex(
    index=teams, columns=season_order, fill_value=0
)
plt.figure(figsize=(14, 9))
sns.heatmap(team_season_wins, annot=True, fmt="g", cmap="Blues")
plt.title("Team Wins Across IPL Seasons")
plt.xlabel("Season")
plt.ylabel("Team")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "07_team_season_wins_heatmap.png", dpi=200, bbox_inches="tight")
plt.show()

# ------------------------------------------------------------
# 8. Toss analysis
# ------------------------------------------------------------
toss_decision_counts = clean["toss_decision"].value_counts()
display(toss_decision_counts)

plt.figure(figsize=(7, 5))
sns.barplot(x=toss_decision_counts.index, y=toss_decision_counts.values)
plt.title("Toss Decision Distribution")
plt.xlabel("Toss Decision")
plt.ylabel("Matches")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "08_toss_decision_distribution.png", dpi=200, bbox_inches="tight")
plt.show()

clean["toss_winner_won_match"] = (
    clean["toss_winner"].eq(clean["winner"]) & clean["winner"].notna()
)

toss_valid = clean[clean["winner"].notna()]
toss_match_rate = toss_valid["toss_winner_won_match"].mean() * 100
print(f"Toss winner also won the match: {toss_match_rate:.2f}%")

toss_result = pd.crosstab(
    toss_valid["toss_winner_won_match"].map({True: "Yes", False: "No"}),
    columns="count"
)
display(toss_result)

plt.figure(figsize=(7, 5))
sns.countplot(data=toss_valid, x="toss_winner_won_match")
plt.title("Toss Winner vs Match Winner")
plt.xlabel("Did Toss Winner Also Win the Match?")
plt.ylabel("Matches")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "09_toss_winner_match_winner.png", dpi=200, bbox_inches="tight")
plt.show()

toss_by_season = pd.crosstab(
    clean["season"], clean["toss_decision"]
).reindex(season_order)
toss_by_season.plot(kind="bar", stacked=True, figsize=(13, 6))
plt.title("Toss Decisions by Season")
plt.xlabel("Season")
plt.ylabel("Matches")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "10_toss_decisions_by_season.png", dpi=200, bbox_inches="tight")
plt.show()

# ------------------------------------------------------------
# 9. Venue analysis
# ------------------------------------------------------------
venue_counts = clean["venue"].value_counts()
top10_venues = venue_counts.head(10).sort_values()

plt.figure(figsize=(11, 7))
sns.barplot(x=top10_venues.values, y=top10_venues.index)
plt.title("Top 10 IPL Venues by Number of Matches")
plt.xlabel("Matches")
plt.ylabel("Venue")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "11_top_venues.png", dpi=200, bbox_inches="tight")
plt.show()

venue_result = pd.crosstab(
    clean["venue"], clean["result"]
).loc[top10_venues.index]
display(venue_result)

# ------------------------------------------------------------
# 10. Result and winning margin analysis
# ------------------------------------------------------------
result_counts = clean["result"].value_counts()
display(result_counts)

plt.figure(figsize=(8, 5))
sns.barplot(x=result_counts.index, y=result_counts.values)
plt.title("Match Result Types")
plt.xlabel("Result Type")
plt.ylabel("Matches")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "12_match_result_types.png", dpi=200, bbox_inches="tight")
plt.show()

# Margin distributions by result
for result_type in ["runs", "wickets"]:
    subset = clean.loc[clean["result"] == result_type, "result_margin"].dropna()
    if len(subset):
        plt.figure(figsize=(9, 5))
        sns.histplot(subset, kde=True)
        plt.title(f"Winning Margin Distribution — {result_type.title()} Wins")
        plt.xlabel("Winning Margin")
        plt.ylabel("Matches")
        plt.tight_layout()
        plt.show()

# ------------------------------------------------------------
# 11. Batting first vs chasing
# ------------------------------------------------------------
# In this dataset, result='runs' represents a successful defense,
# while result='wickets' represents a successful chase.
bat_first_success = clean["result"].eq("runs")
chase_success = clean["result"].eq("wickets")

outcome_type = pd.Series(
    np.select(
        [bat_first_success, chase_success],
        ["Defended (won by runs)", "Chased (won by wickets)"],
        default="Tie/No Result"
    ),
    index=clean.index
)

outcome_counts = outcome_type.value_counts()
display(outcome_counts)

plt.figure(figsize=(9, 5))
sns.barplot(x=outcome_counts.values, y=outcome_counts.index)
plt.title("Match Outcome Type")
plt.xlabel("Matches")
plt.ylabel("Outcome")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "13_defended_vs_chased.png", dpi=200, bbox_inches="tight")
plt.show()

# ------------------------------------------------------------
# 12. Head-to-head analysis
# ------------------------------------------------------------
# Build unordered pair and count wins for each team.
h2h = pd.DataFrame(0, index=teams, columns=teams, dtype=int)

for _, row in clean.dropna(subset=["winner"]).iterrows():
    t1, t2, winner = row["team1"], row["team2"], row["winner"]
    if winner in h2h.index and t1 in h2h.index and t2 in h2h.columns:
        h2h.loc[winner, t1 if winner != t1 else t2] += 1

plt.figure(figsize=(13, 10))
sns.heatmap(h2h, annot=True, fmt="g", cmap="Blues")
plt.title("Head-to-Head Wins by Team")
plt.xlabel("Opponent")
plt.ylabel("Winning Team")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "14_head_to_head_heatmap.png", dpi=200, bbox_inches="tight")
plt.show()

# ------------------------------------------------------------
# 13. Super-over and D/L analysis
# ------------------------------------------------------------
super_counts = clean["super_over"].value_counts()
print("Super-over matches:", int(super_counts.get("Y", 0)))

plt.figure(figsize=(6, 5))
sns.barplot(x=super_counts.index, y=super_counts.values)
plt.title("Super Over Distribution")
plt.xlabel("Super Over")
plt.ylabel("Matches")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "15_super_over_distribution.png", dpi=200, bbox_inches="tight")
plt.show()

method_counts = clean["method"].fillna("No D/L").value_counts()
display(method_counts)

# ------------------------------------------------------------
# 14. Correlation analysis
# ------------------------------------------------------------
corr_cols = [
    c for c in ["result_margin", "target_runs", "target_overs"]
    if c in clean.columns
]
corr = clean[corr_cols].corr()
display(corr)

plt.figure(figsize=(7, 5))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Heatmap of Numeric Match Variables")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "16_correlation_heatmap.png", dpi=200, bbox_inches="tight")
plt.show()

# ------------------------------------------------------------
# 15. Outlier detection using IQR
# ------------------------------------------------------------
def iqr_summary(data, col):
    s = data[col].dropna()
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    outliers = s[(s < lower) | (s > upper)]
    return {
        "column": col,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "lower_bound": lower,
        "upper_bound": upper,
        "outlier_count": len(outliers),
        "outlier_percentage": len(outliers) / len(s) * 100
    }

outlier_table = pd.DataFrame([
    iqr_summary(clean, c) for c in numerical_cols
])
display(outlier_table)

# ------------------------------------------------------------
# 16. Important summary statistics
# ------------------------------------------------------------
summary_stats = {
    "Total matches": len(clean),
    "Seasons": clean["season"].nunique(),
    "Teams represented": len(set(clean["team1"]) | set(clean["team2"])),
    "Venues": clean["venue"].nunique(),
    "Cities": clean["city"].nunique(),
    "Toss decisions — field": int((clean["toss_decision"] == "field").sum()),
    "Toss decisions — bat": int((clean["toss_decision"] == "bat").sum()),
    "Toss winner also match winner (%)": round(toss_match_rate, 2),
    "Wins by runs": int((clean["result"] == "runs").sum()),
    "Wins by wickets": int((clean["result"] == "wickets").sum()),
    "Ties": int((clean["result"] == "tie").sum()),
    "No results": int((clean["result"] == "no result").sum()),
    "Super overs": int((clean["super_over"] == "Y").sum()),
    "D/L matches": int((clean["method"] == "D/L").sum()),
}
summary_table = pd.DataFrame(
    {"Metric": summary_stats.keys(), "Value": summary_stats.values()}
)
display(summary_table)

summary_table.to_csv(OUTPUT_DIR / "summary_statistics.csv", index=False)

# ------------------------------------------------------------
# 17. Automatic data-driven insights
# ------------------------------------------------------------
top_wins_team = team_summary["wins"].idxmax()
top_wins = int(team_summary.loc[top_wins_team, "wins"])

most_matches_team = team_summary["matches_played"].idxmax()
most_matches = int(team_summary.loc[most_matches_team, "matches_played"])

most_common_venue = venue_counts.idxmax()
most_common_venue_matches = int(venue_counts.max())

most_common_result = result_counts.idxmax()
most_common_result_count = int(result_counts.max())

highest_margin_runs = clean.loc[
    clean["result"].eq("runs"), "result_margin"
].max()
highest_margin_wickets = clean.loc[
    clean["result"].eq("wickets"), "result_margin"
].max()

print("KEY DATA-DRIVEN OBSERVATIONS")
print(f"1. The dataset contains {len(clean):,} matches across {clean['season'].nunique()} season labels.")
print(f"2. {top_wins_team} has the highest recorded match-win count: {top_wins}.")
print(f"3. {most_matches_team} appears in the highest number of matches: {most_matches}.")
print(f"4. {most_common_venue} hosts the highest number of matches in this dataset: {most_common_venue_matches}.")
print(f"5. The most common result type is '{most_common_result}' with {most_common_result_count} matches.")
print(f"6. Toss winners also won the match in {toss_match_rate:.2f}% of matches with a recorded winner.")
print(f"7. Matches won by runs: {(clean['result'] == 'runs').sum()}.")
print(f"8. Matches won by wickets: {(clean['result'] == 'wickets').sum()}.")
print(f"9. Tied matches: {(clean['result'] == 'tie').sum()}; no-result matches: {(clean['result'] == 'no result').sum()}.")
print(f"10. Super-over matches: {(clean['super_over'] == 'Y').sum()}; D/L matches: {(clean['method'] == 'D/L').sum()}.")

if pd.notna(highest_margin_runs):
    print(f"11. Largest recorded run-margin victory: {highest_margin_runs:.0f} runs.")
if pd.notna(highest_margin_wickets):
    print(f"12. Largest recorded wicket-margin victory: {highest_margin_wickets:.0f} wickets.")

# ------------------------------------------------------------
# 18. Save cleaned dataset
# ------------------------------------------------------------
clean.to_csv(OUTPUT_DIR / "ipl_matches_cleaned.csv", index=False)

print("\nEDA completed.")
print(f"Charts and outputs saved to: {OUTPUT_DIR.resolve()}")
```
